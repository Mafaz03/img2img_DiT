```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
with open("../img2img_DiT.out", encoding="utf-8") as f:
    content = f.readlines()
```

```python
# First occurrence
for idx_start, line in enumerate(content):
    if line.startswith("[DiT]"):
        break

# Last occurrence
for rev_idx, line in enumerate(reversed(content)):
    if line.startswith("[DiT]"):
        idx_end = len(content) - rev_idx
        break
```

```python
diff_loss  = []
clip_loss  = []
lpips_loss = []

for l in content[idx_start:idx_end]:
    diff_loss.append(float(l.split("=")[2].split(" ")[0]))
    clip_loss.append(float(l.split("=")[3].split(" ")[0]))
    lpips_loss.append(float(l.split("=")[4].split(" ")[0]))
    
```

```python
plt.plot(diff_loss, label = "Diffusion loss")
plt.plot(clip_loss, label = "Clip loss")
plt.plot(lpips_loss, label = "Lpips loss")

plt.legend()
plt.grid()
plt.show()
```
